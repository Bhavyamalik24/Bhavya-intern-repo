# Focus Bear User Analysis
## Introduction
This notebook explores simulated Focus Bear user data to understand
usage patterns, habit completion rates, and focus session trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python
data = {
    "user_id": range(1, 21),
    "days_active": [5, 12, 3, 20, 8, 15, 1, 25, 7, 18,
                    10, 22, 4, 16, 9, 30, 2, 14, 6, 28],
    "habits_completed": [3, 8, 1, 15, 5, 10, 0, 18, 4, 12,
                         7, 16, 2, 11, 6, 20, 1, 9, 3, 19],
    "focus_minutes": [25, 60, 10, 90, 40, 75, 5, 100, 35, 80,
                      50, 85, 15, 70, 45, 120, 8, 65, 30, 110],
    "membership": ["free", "premium", "free", "premium", "free",
                   "premium", "free", "premium", "free", "premium",
                   "free", "premium", "free", "premium", "free",
                   "premium", "free", "premium", "free", "premium"]
}

df = pd.DataFrame(data)
print(f"Dataset shape: {df.shape}")
df.head()
```

## Dataset Overview
The dataset contains 20 simulated Focus Bear users with the following fields:
- **user_id**: Unique identifier
- **days_active**: Number of days the user has been active
- **habits_completed**: Total habits completed
- **focus_minutes**: Average daily focus minutes
- **membership**: Free or Premium tier

```python
# Basic statistics
df.describe()
```

```python
# Focus minutes by membership type
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="membership", y="focus_minutes", palette="Set2")
plt.title("Focus Minutes by Membership Type")
plt.xlabel("Membership")
plt.ylabel("Focus Minutes")
plt.show()
```

```python
# Habits completed vs days active
plt.figure(figsize=(8, 4))
sns.scatterplot(data=df, x="days_active", y="habits_completed",
                hue="membership", size="focus_minutes",
                sizes=(50, 200), palette="Set1")
plt.title("Habits Completed vs Days Active")
plt.xlabel("Days Active")
plt.ylabel("Habits Completed")
plt.show()
```

## Key Insights
- Premium users consistently complete more habits and focus for longer
- There is a clear positive correlation between days active and habits completed
- New users (fewer days active) show lower engagement regardless of membership
